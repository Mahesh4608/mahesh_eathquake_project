```python
%py
from databricks.sdk import WorkspaceClient
w = WorkspaceClient()
conn = w.connections.get("mahesh_earthquake_conn")
print(conn)
base_url = f"{conn.options['host']}{conn.options['base_path']}"
```

```python
import requests
import json
import datetime

url = f"{base_url}/summary/all_day.geojson"
response = requests.get(url)
if response.status_code != 200:
    raise Exception(f"error in getting data from {url}")
data = response.json()
current_date = datetime.datetime.now().strftime("%Y-%m-%d")
dbutils.fs.put(
    f"/Volumes/{catalog_name}/bronze/earthqauke_data/earthqauke_data_{current_date}.json",
    json.dumps(response.json()),
    overwrite=True,
)
```
